```python
import tensorflow as tf
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.applications import EfficientNetB0
from tensorflow.keras import layers, models
```

```python
dataset_path = "dataset"
```

```python
train_datagen = ImageDataGenerator(
    rescale=1./255,
    rotation_range=20,
    zoom_range=0.2,
    horizontal_flip=True,
    validation_split=0.2
)

train_generator = train_datagen.flow_from_directory(
    dataset_path,
    target_size=(224, 224),
    batch_size=32,
    class_mode='categorical',
    subset='training'
)

val_generator = train_datagen.flow_from_directory(
    dataset_path,
    target_size=(224, 224),
    batch_size=32,
    class_mode='categorical',
    subset='validation'
)
```

```python
base_model = EfficientNetB0(
    weights='imagenet',
    include_top=False,
    input_shape=(224,224,3)
)

base_model.trainable = False

model = models.Sequential([
    base_model,
    layers.GlobalAveragePooling2D(),
    layers.Dense(128, activation='relu'),
    layers.Dense(4, activation='softmax')
])
```

```python
model.compile(
    optimizer='adam',
    loss='categorical_crossentropy',
    metrics=['accuracy']
)
```

```python
# 1. Define the importance of each class
# We give Puffy Eyes a very low weight (0.1) so the AI stops guessing it
custom_weights = {
    0: 5.0,  # Clear Skin (Study 5x harder!)
    1: 4.0,  # Dark Spots
    2: 0.1,  # Puffy Eyes (Already knows this, lower priority)
    3: 3.0   # Wrinkles
}

# 2. Re-compile with a slow learning rate for better accuracy
from tensorflow.keras.optimizers import Adam
model.compile(
    optimizer=Adam(learning_rate=0.00001), 
    loss='categorical_crossentropy', 
    metrics=['accuracy']
)

# 3. Start the long training (30 epochs)
history = model.fit(
    train_generator,
    epochs=30,
    class_weight=custom_weights,
    validation_data=val_generator
)

# 4. Save the "Smart" version of your model
model.save('skin_aging_model.h5')
print("Success! The new balanced model is saved.")
```
